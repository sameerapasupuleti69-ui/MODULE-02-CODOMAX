```python

import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

print("--- Dataset First 5 Rows ---")
print(df.head())

print("\n--- Dataset Info ---")
print(df.info())

print("\n--- Dataset Shape ---")
print("Rows:", df.shape[0], "Columns:", df.shape[1])

print("\n--- Basic Statistics ---")
print(df.describe())

print("\n--- Mean Total Bill (NumPy) ---")
print(np.mean(df['total_bill']))

print("\n--- Max Tip ---")
print(np.max(df['tip']))

print("\n--- Missing Values ---")
print(df.isnull().sum())

df.loc[0, 'total_bill'] = np.nan
print("\nAfter making 1 value missing:")
print(df.isnull().sum())


df['total_bill'].fillna(df['total_bill'].mean(), inplace=True)
print("\nAfter Filling Missing Values:")
print(df.isnull().sum())

print("\nDuplicates before:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after:", df.duplicated().sum())
df['tip_percentage'] = (df['tip'] / df['total_bill']) * 100
print("\n--- Final Cleaned Data ---")
print(df.head(10))

print("\nModule 02 Completed Successfully!")
```
